```python
import pandas as pd
import numpy as np
import pandas as pd
import missingno as msno
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/brain_tumor_data.csv')
```

Load Data   

```python
#df.describe()
df.isnull().sum()

```

```python



# 1. Check if lifestyle columns are missing together
lifestyle = ['bmi', 'smoking_status', 'alcohol_consumption']
print("=== LIFESTYLE MISSING PATTERN ===")
print(df[lifestyle].isnull().sum())

# How many patients have ALL three missing?
all_three_missing = df[lifestyle].isnull().all(axis=1).sum()
print(f"All three lifestyle missing: {all_three_missing}")

# 2. Correlation of missing indicators (1=missing, 0=not missing)
missing_corr = df[lifestyle].isnull().corr()
print("\n=== MISSING VALUE CORRELATIONS ===")
print(missing_corr)

# 3. Visualize missing patterns
msno.matrix(df[lifestyle])
plt.title("Lifestyle Columns Missing Pattern")
plt.show()

# 4. Full dataset heatmap
msno.heatmap(df)
plt.title("All Columns Missing Value Correlations")
plt.show()
```
